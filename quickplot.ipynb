```python
import matplotlib.pyplot as plt
import numpy as np

with np.load('out/output.npz') as f:
    print(f.files)
    time = f['t']
    states = f['x']
    control = f['u']
    command = f['c']
    wind = f['w']
    aero = f['a']
```

```python
fig, axs = plt.subplots(3,3, figsize=(12,12))
i = 0
for ax in axs.flat:
    ax.grid(True)

    if i == 0:
        ax.plot(time, states[:,6], label='roll')
        ax.plot(time, command[:,3], label='roll command')
        ax.set_xlabel('Time (s)')
        ax.set_ylabel('Roll (rad)')
        ax.set_title('Roll')
        ax.legend()

    if i == 1:
        ax.plot(time, states[:,7], label='pitch')
        ax.plot(time, command[:,4], label='pitch command')
        ax.set_xlabel('Time (s)')
        ax.set_ylabel('Pitch (rad)')
        ax.set_title('Pitch')
        ax.legend()

    if i == 2:
        ax.plot(time, states[:,8], label='yaw')
        ax.plot(time, command[:,5], label='yaw command')
        ax.set_xlabel('Time (s)')
        ax.set_ylabel('Yaw (rad)')
        ax.set_title('Yaw')
        ax.legend()

    if i == 3:
        ax.plot(time, states[:,9], label='p')
        ax.plot(time, command[:,0], label='p command')
        ax.set_xlabel('Time (s)')
        ax.set_ylabel('Roll rate (rad/s)')
        ax.set_title('Roll Rate')
        ax.legend()

    if i == 4:
        ax.plot(time, states[:,10], label='q')
        ax.plot(time, command[:,1], label='q command')
        ax.set_xlabel('Time (s)')
        ax.set_ylabel('Pitch rate (rad/s)')
        ax.set_title('Pitch Rate')
        ax.legend()

    if i == 5:
        ax.plot(time, states[:,11], label='r')
        ax.plot(time, command[:,2], label='r command')
        ax.set_xlabel('Time (s)')
        ax.set_ylabel('Yaw rate (rad/s)')
        ax.set_title('Yaw Rate')
        ax.legend()

    if i == 6:
        ax.plot(time, control[:,0])
        ax.set_xlabel('Time (s)')
        ax.set_ylabel('Elevator Position (rad)')
        ax.set_title('Elevator')

    if i == 7:
        ax.plot(time, control[:,2])
        ax.set_xlabel('Time (s)')
        ax.set_ylabel('Aileron Position (rad)')
        ax.set_title('Aileron')

    if i == 8:
        ax.plot(time, control[:,3])
        ax.set_xlabel('Time (s)')
        ax.set_ylabel('Rutter Position (rad)')
        ax.set_title('Rutter')


    i+=1
plt.tight_layout()
```
